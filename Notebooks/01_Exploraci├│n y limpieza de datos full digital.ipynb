Se importa la libreria pandas para el manejo de datos y se cargan los datos originales CSV.

## Duplicated Value Verification

## ID´s Validation

Se verifica que los ID´s sean válidos y consistentes.

```python
print("\nID ruta minimo")
print(full_digital_df['Ruta SAP'].min())

# ------------------------------------------------

print("\nID cliente minimo")
print(full_digital_df['Cliente ID'].min())
```

## Names Data Cleaning

Se estandarizan los nombres de clientes para consistencia

```python
full_digital_df['Nombre Cliente'] = full_digital_df['Nombre Cliente'].str.strip()   # Quitar espacios
full_digital_df['Nombre Cliente'] = full_digital_df['Nombre Cliente'].str.title()   # Estandarizar mayúsculas
print("\nPrimeros 5 nombres normalizados:")
print(full_digital_df['Nombre Cliente'].head())
```

## Status Data Cleaning

Se estandarizó la columna 'Categoría ' y se verificó que solo contenga 'non digital', 'hibrido' o 'fully'.

```python
full_digital_df['Categoría'] = full_digital_df['Categoría'].str.lower().str.strip()
print("\nValores únicos en Status Registro:")
print(full_digital_df['Categoría'].unique())
```

## Standardization of Column Headers

Se estandariza el nombre de los encabezados de los campos

```python
full_digital_df.columns = (
    full_digital_df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

full_digital_df.head()
```

## Dataset clean and ready to export

Dataset exportado a CSV para uso en análisis y joins futuros.

```python
print("\nDataset limpio y listo para análisis")
full_digital_df.to_csv(r"../Data_processed_anon/Fully_digital_clean.csv", index=False)
```
