## Import Libraries and Data Load

Se importa la libreria pandas para el manejo de datos y se cargan los datos procesados CSV.

```python
import pandas as pd

status_df = pd.read_csv(r"../Data_processed_anon/Fully_digital_clean.csv")
registros_df= pd.read_csv(r"../Data_processed_anon/Registros_Clientes_clean.csv")
```

## Client ID Mapping

Se crea un mapeo para el campo Cliente ID.

```python
#Se construye el universo de clientes

all_clients = pd.concat([
    registros_df['nip_cliente'],
    status_df['cliente_id']
]).unique()

```

```python
#Se construye el mapa
client_map = {
    original_id: f"CUST_{str(i+1).zfill(4)}"
    for i, original_id in enumerate(all_clients)
}
```

```python
#Se mapea 
registros_df['cliente_anon_id'] = registros_df['nip_cliente'].map(client_map)
status_df['cliente_anon_id']    = status_df['cliente_id'].map(client_map)
```

## Route ID Mapping

Se crea un mapeo para el campo Ruta ID.

```python
#Se construye el universo de rutas

all_routes = pd.concat([
    registros_df['ruta'],
    status_df['ruta_sap']
]).unique()
```

```python
#Se construye el mapa
route_map = {
    original_id: f"ROUTE_{str(i+1).zfill(4)}"
    for i, original_id in enumerate(all_routes)
}
```

```python
registros_df["ruta_anon"] = registros_df["ruta"].map(route_map)
status_df["ruta_anon"] = status_df["ruta_sap"].map(route_map)
```

## Name Anonymized

Se anonimizarón los nombre para proteger la confidencialidad e información sensible.

```python
registros_df[['cliente_anon_id', 'nombre_cliente']].head()
```

```python
unique_clients = registros_df['cliente_anon_id'].dropna().unique()

name_map = {
    client: f"Store_{str(i+1).zfill(4)}"
    for i, client in enumerate(unique_clients)
}
```

```python
status_df['nombre_anon'] = status_df['cliente_anon_id'].map(name_map)
registros_df['nombre_anon'] = registros_df['cliente_anon_id'].map(name_map)
```

## Fields Dropped

```python
registros_df = registros_df.drop(columns=["distrito_sap", "nip_cliente", "nombre_cliente", "ruta"])

new_order = [
    "ruta_anon",
    "cliente_anon_id",
    "nombre_anon",
    "status_registro",
   
]

registros_df = registros_df[new_order]
registros_df.head()
```

```python
status_df = status_df.drop(columns=["región", "división", "distrito", "cedis", "ruta_sap", "cliente_id", "nombre_cliente", "venta_digital"])

new_order = [
    "ruta_anon",
    "cliente_anon_id",
    "nombre_anon",
    "categoría",
    "%_venta_digital",
   
]

status_df = status_df[new_order]
status_df.head()
```
