## Names Data Cleaning

Se estandarizan los nombres de clientes para consistencia

```python
registros_clientes_df['Nombre Cliente'] = registros_clientes_df['Nombre Cliente'].str.strip()   # Quitar espacios
registros_clientes_df['Nombre Cliente'] = registros_clientes_df['Nombre Cliente'].str.title()   # Estandarizar mayúsculas
print("\nPrimeros 5 nombres normalizados:")
print(registros_clientes_df['Nombre Cliente'].head())
```

## Status Data Cleaning

Se estandarizó la columna 'Categoría ' y se verificó que solo contenga 'non digital', 'hibrido' o 'fully'.

```python
registros_clientes_df['Status Registro'] = registros_clientes_df['Status Registro'].str.lower().str.strip()
print("\nValores únicos en Status Registro:")
print(registros_clientes_df['Status Registro'].unique())
```

```python
registros_clientes_df.head(10)
```

## Standardization of Column Headers

Se estandariza el nombre de los encabezados de los campos

```python
registros_clientes_df.columns = (
    registros_clientes_df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

registros_clientes_df.head()
```

## Dataset clean and ready to export

Dataset exportado a CSV para uso en análisis y joins futuros.

```python
print("\nDataset limpio y listo para análisis")
registros_clientes_df.to_csv(r"../Data_processed_anon/Registros_Clientes_clean.csv", index=False)
```
